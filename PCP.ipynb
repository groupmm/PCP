# Introduction

The Preparation Course for Python (PCP) is an open-source course in scientific programming built around ten Jupyter notebooks. The opening units cover the working environment, Python, NumPy, control structures, functions, and visualization. The later units apply these tools to complex numbers, exponential functions, signals, sampling, the discrete Fourier transform (DFT), and the fast Fourier transform (FFT). A final unit introduces modules and packages for organizing reusable code.

PCP concentrates on a practical selection of Python concepts rather than attempting to cover the entire language. Some general programming experience and basic knowledge of digital signal processing are helpful, but the essential mathematical ideas are reviewed as they arise. The notebooks support self-study, undergraduate programming and laboratory courses, and preparation for graduate study. We use them in teaching and research internships connected with international Master’s programs, including [Communications and Multimedia Engineering (CME)](https://www.cme.studium.fau.de/), [Electrical Engineering (EEI)](https://www.eei.studium.fau.de/), [Information and Communication Technology (ICT)](https://www.iuk.studium.fau.eu/), [Data Science](https://www.datascience.nat.fau.eu/), and [Artificial Intelligence (AI)](https://www.aibe.tf.fau.de/).

The PCP course follows a simple learning structure:

* **Read and run**: Explanations, formulas, and executable code alternate, so each idea can be tested immediately.

* **Explore the context**: Green boxes provide mathematical background, signal-processing connections, applications, and occasional excursions beyond the main topic.

* **Notice the details**: Orange boxes state learning objectives, offer practical guidance, and point out common pitfalls.

* **Put ideas into practice**: Each unit concludes with exercises in blue boxes, giving readers an opportunity to develop and test their own solutions. Sample solutions are available through the accompanying `libpcp` package.

* **Develop good scientific practices**: Conda, Jupyter, Colab, Binder, and [GitHub](https://github.com/meinardmueller/PCP) introduce readers to managing environments, organizing code, and working reproducibly.

PCP also provides a foundation for the [Preparation Course for PyTorch (PCPT)](https://www.audiolabs-erlangen.de/PCPT) and the [Fundamentals of Music Processing (FMP)](https://www.audiolabs-erlangen.de/FMP) notebooks. Readers may follow the units in order or return to individual topics as needed. For the first pass, however, working through the examples and exercises sequentially is recommended.

## Get Started

If a static view of the PCP notebooks is enough for you, the [exported HTML versions](https://audiolabs-erlangen.de/PCP) can be used right away without any installation. All material, including the explanations, the figures, and the audio examples, can be accessed by just following the **HTML links**. If you want to **execute** the Python code cells, you have to download the notebooks (along with the data), create an environment, and start a Jupyter server. You then need to follow the **IPYNB links** within the Jupyter <span style="hyphens: none">session</span>. As an alternative for executing the notebooks, one can also use web-based services such as Google colab or binder via  the [notebooks' GitHub repository](https://github.com/groupmm/PCP). The necessary steps and technical details are explained in the [PCP notebook Get Started](PCP_01_getstarted.html).

## Structure and Content

The PCP collection comprises ten units, each presented as an individual notebook. Every unit begins with an overview and its learning objectives, followed by executable code, textbook-style explanations, mathematical formulas, plots, images, and sound examples. Short coding exercises conclude each unit. For self-evaluation, a sample solution to every exercise is implemented as a Python function in the corresponding module of the [Python package `libpcp`](PCP_10_module.html). The opening units establish the foundations of scientific programming, while the later units introduce the required mathematics, signals, and sampling before arriving at the DFT and FFT. The final unit explains how code can be organized into reusable modules and packages. The following table summarizes the ten units and their main topics.

## Further Notes, Links, and Literature

The PCP notebooks provide the Python foundations needed for more advanced programming and laboratory courses, particularly in signal processing. They are not intended as a comprehensive introduction to Python, such as [The Python Tutorial](https://docs.python.org/3/tutorial/index.html) or the [SciPy Lecture Notes](https://scipy-lectures.org/). Instead, they focus on essential data structures, numerical computing, visualization, and basic functions while leaving out advanced topics such as object-oriented programming. This focused scope helps beginners develop practical skills without being overwhelmed.

Alongside programming, a central educational goal of PCP is to review the mathematical foundations of Fourier analysis and make the FFT more accessible. Rather than presenting the FFT as an isolated clever algorithm, the notebooks build toward it through complex numbers, exponential functions, signals, sampling, and the DFT. Executable code, visualizations, and sound examples help turn these abstract ideas into concrete experiments.

PCP also prepares readers for more advanced tools and learning materials. The [FMP notebooks](https://www.audiolabs-erlangen.de/FMP) build on these foundations to teach methods for analyzing and processing music signals. The [PCPT notebooks](https://www.audiolabs-erlangen.de/PCPT) continue the path toward PyTorch, neural networks, and modern machine learning from a signal-processing perspective. We also recommend [librosa](https://librosa.org/doc/latest/index.html), a widely used Python package offering flexible implementations of standard methods in audio and music processing.

In summary, the PCP notebooks bridge the gap between programming and mathematical thinking, helping students develop a conceptual and practical understanding of scientific Python and Fourier analysis.
