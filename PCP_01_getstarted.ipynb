## Python, Miniconda, and Conda

The <strong>Preparation Course for Python</strong> introduces Python through examples from scientific computing, mathematics, and signal processing. Before we begin learning the language in <a href="PCP_02_python.html">Unit 2</a>, we first prepare the software needed to run the examples and complete the exercises. This includes Python itself and several additional packages used throughout the course. The basic Python language does not provide every tool needed for scientific computing. Additional collections of code, called <strong>packages</strong>, extend its functionality. PCP uses packages such as **NumPy** for numerical calculations, **Matplotlib** for visualization, and **Jupyter Notebook** for interactive computing. Some packages require other packages in order to work. These requirements are called <strong>dependencies</strong>.

Installing Python and managing all required packages individually would be inconvenient. We therefore use <a href="https://docs.conda.io/"><strong>Conda</strong></a>, a tool for managing packages and Python environments. Conda determines which dependencies are required, installs suitable versions, and keeps the software used by different projects separate. <a href="https://docs.conda.io/projects/conda/en/stable/user-guide/install/index.html"><strong>Miniconda</strong></a> is a small distribution that provides Python, Conda, and a few essential components. It gives us everything needed to create the PCP environment without installing a large collection of unrelated packages. To install Miniconda:

<ol>
  <li>Open the official <a href="https://docs.conda.io/projects/conda/en/stable/user-guide/install/index.html">Conda installation page</a>.</li>
  <li>Select the Miniconda installer for your operating system.</li>
  <li>Run the installer and follow its instructions.</li>
  <li>On Windows, open the <strong>Anaconda Prompt</strong>. On macOS or Linux, open a terminal application.</li>
  <li>Verify the installation by entering: <code>conda --version</code></li>
</ol>

If Conda is installed correctly, the final command displays its version number. The exact number may differ between computers. We can now use Conda to create a separate software environment for PCP.

## Python Environments and the PCP Environment File

A Python project rarely uses Python alone. It may depend on many packages, and these packages continue to change as developers correct errors and add new features. Different projects may therefore require different versions of the same package. If all packages are installed together, updating the software for one project may cause another project to stop working. A <strong>Python environment</strong> keeps the software for one project separate from the software used by other projects. You can think of it as a dedicated toolbox containing a particular Python version and a particular collection of packages. Activating an environment tells the computer which toolbox to use. PCP therefore has its own environment containing the software versions intended for this course.

It is important to distinguish the <strong>project directory</strong> from the <strong>Python environment</strong>. The project directory contains the files you work with, including notebooks, Python modules, images, audio, and data. The environment contains the software used to open and execute those files. Deleting an environment does not delete the project files, and copying the project directory does not automatically install the required software.

The PCP project includes a file named <code>environment.yml</code>. The extension <code>.yml</code> indicates that the file uses <a href="https://yaml.org/"><strong>YAML</strong></a>, a text format for storing structured information in a form that is easy for both people and programs to read. YAML files are often used to describe the settings and software requirements of a project. In this case, <code>environment.yml</code> specifies the name of the PCP environment, the locations from which packages are obtained, and the software required by the course. The following code cell reads the file and displays its current contents. You do not need to understand the Python code yet; its purpose here is simply to show the file.

```python
fn = os.path.join('environment.yml')

with open(fn, 'r', encoding='utf-8') as stream:
    env = stream.read()

print(env)
```

The first entry gives the environment the name <code>PCP</code>. The <code>channels</code> section tells Conda where packages should be downloaded. In this case, the packages come from <code>conda-forge</code>, a widely used community collection of software packages. The entry <code>nodefaults</code> tells Conda not to search its standard package channels in addition to the listed channel. The <code>dependencies</code> section lists the software to be installed in the environment. An entry such as <code>python=3.14.*</code> specifies both a package and an allowed version range. The asterisk permits any available update within version <code>3.14</code>. The final subsection lists packages installed using <strong>pip</strong>, Python's own package installer. Indentation is part of the structure of a YAML file, so indented entries belong to the section above them.

You do not need to install these packages individually or decide which versions work together. This is exactly what Conda is designed to manage. It reads <code>environment.yml</code>, identifies the requested packages and their dependencies, and searches for a collection of versions that are compatible with one another. This process is called <strong>dependency resolution</strong>. Conda then downloads the selected packages and installs them together in the new environment. A <strong>conflict</strong> occurs when no collection of package versions can satisfy all requirements. For example, one package may require an older version of another package, while a second package requires a newer and incompatible version. In this situation, Conda reports the problem instead of silently creating an environment whose packages cannot work together. Specifying the package versions also supports <strong>reproducibility</strong>: students working on different computers can create closely matching software environments and are therefore more likely to obtain the same results.

## Using Jupyter Notebooks

The PCP course is presented through interactive documents called **Jupyter notebooks**. These notebooks are opened and edited using [**Jupyter Notebook**](https://jupyter-notebook.readthedocs.io/en/latest/), an application designed for interactive computing. A notebook is divided into sections called **cells**. A **Markdown cell** contains explanatory text, headings, lists, links, images, and mathematical formulas, while a **code cell** contains executable Python instructions and displays their results directly below the cell. Together, these cells allow explanations, calculations, figures, and audio to appear in a single document. PCP uses this format to place explanations and experiments next to each other. You can read a mathematical idea, execute the corresponding code, change a parameter, and immediately see or hear the result. The examples are therefore not fixed demonstrations. They are starting points that you can explore, modify, and extend with your own ideas.

A notebook is usually read and executed from top to bottom. Select a cell and press <code>Shift + Enter</code> to process it and continue to the next cell. For a **Markdown cell**, this turns its source text into formatted content. For a **code cell**, it sends the Python instructions to the <strong>kernel</strong>, the running Python process that performs the computations. A number such as <code>[3]</code> beside a code cell indicates when it was executed during the current session. Empty brackets <code>[ ]</code> indicate that the cell has not yet been executed, while <code>[*]</code> indicates that it is still running. The kernel remembers imported packages, variables, functions, and previous results until it is restarted or stopped. This remembered information is called the notebook's **state**. A later cell can use a variable created in an earlier cell, but only if the earlier cell has already been executed. Since cells can be run in any order, the visible order of a notebook does not always match its actual execution order.

Before trusting or exporting a notebook, restart the kernel and execute all cells from beginning to end. This checks that every result follows from the visible notebook content rather than from an earlier experiment.

## Exporting Notebooks

A Jupyter notebook is an interactive document whose code cells can be edited and executed. It can also be exported as a **static document** that records its current text, code, figures, and saved results. An **HTML export** can be opened in a web browser and is well suited for online reading, while a **PDF export** is convenient for printing and sharing. The exported document can be viewed but its code cannot be executed interactively. You should therefore keep the original `.ipynb` file as the editable and executable version.

Before exporting a notebook, save it, restart the kernel, and execute all cells from beginning to end. This checks that the displayed results can be reproduced from the notebook itself and do not depend on cells executed earlier or out of order. To export through Jupyter Notebook, select <strong>File &rarr; Save and Export Notebook As</strong> and choose the desired format. Notebooks can also be processed from a terminal using [**nbconvert**](https://nbconvert.readthedocs.io/), a command-line tool provided by Jupyter.
