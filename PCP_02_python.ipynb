## Variables and Basic Operators

A **variable** gives a name to a value, allowing us to store information and use it again later. For example, the statement `year = 2026` stores the value `2026` under the name `year`. The symbol `=` is called the **assignment operator**. It assigns the value on the right to the variable on the left, so `year = 2026` means assigning the value `2026` to `year`. Here are a few basic rules for variable names:

* A variable is created when a value is first assigned to it.
* A variable and its type do not need to be declared beforehand.
* A variable name may contain letters, digits, and underscores (`_`).
* A variable name must begin with a letter or an underscore, not with a digit.
* Names are **case-sensitive**, so `year`, `Year`, and `YEAR` refer to three different variables.
* Variable names usually use lowercase letters, with underscores separating words, as in `sampling_rate`.

Text is represented by a **string**. A string is enclosed in either single quotes (`'`) or double quotes (`"`). Both forms work in the same way. In this course, we usually use single quotes for consistency. The following example assigns a string to a variable and displays it using the `print` function.

```python
welcome_message = 'Welcome to the Preparation Course for Python'
print(welcome_message)
```

Python can perform calculations directly. The following example assigns the value `3` to `n`, applies several basic arithmetic operators, and prints the results:

```python
n = 3                         # Assign the value 3 to n
print('n + 1 =', n + 1)       # Addition
print('n - 1 =', n - 1)       # Subtraction
print('n * 2 =', n * 2)       # Multiplication
print('n / 2 =', n / 2)       # Division
print('n ** 2 =', n ** 2)     # Exponentiation
```

In **Python 3**, the division operator `/` always returns a floating-point number, even when the division has no remainder. The operator `//` performs integer division, which is useful when an integer value is needed, for example as an index. The operator `%` returns the remainder. Python 2 treats division differently, but this course uses Python 3.

```python
n = 8
print('n / 2  =', n / 2)    # Division: 4.0
print('n // 2 =', n // 2)   # Integer division: 4
print('n / 5  =', n / 5)    # Division: 1.6
print('n // 5 =', n // 5)   # Integer division: 1
print('n % 5  =', n % 5)    # Remainder: 3
```

A variable can be updated using its current value. Python provides compact assignment operators that combine a calculation with an assignment. For example, `n += 11` is a shorter form of `n = n + 11`. Each operation updates `n`, so the following operation uses the new value.

```python
n = 7
n += 11                              # Add 11 to n
print('After addition:', n)
n *= 2                               # Multiply n by 2
print('After multiplication:', n)
n /= 18                              # Divide n by 18
print('After division:', n)
n **= 0.5                            # Raise n to the power 0.5
print('After exponentiation:', n)
```

For more control over printed text, Python provides [formatted string literals](https://docs.python.org/3/tutorial/inputoutput.html#formatted-string-literals), usually called **f-strings**. An f-string begins with `f`, and expressions inside braces `{}` are evaluated and inserted into the text. Optional formatting instructions can control the width and precision of a number. Here are a few examples:

```python
number = 17
text = 'ABCD'
pi = 3.14159265359

print(f'Printing a number: {number}')                            # Insert a number
print(f'Printing a string: {text}')                              # Insert a string
print(f'Integer using a field of width 6: {1234:6d}')            # Minimum width: 6
print(f'Floating-point number with 3 decimal places: {pi:.3f}')  # Precision: 3
```

## Lists and Tuples

**Lists** and **tuples** allow several values to be collected in a specific order. They may contain different kinds of values, including numbers, strings, and even other lists or tuples. A list is written using square brackets, such as `[2, 4, 6]`, while a tuple uses parentheses, such as `(2, 4, 6)`. Individual elements are accessed using square brackets. Python starts counting at `0`, so the first element has index `0`, the second has index `1`, and so on. The function `len()` returns the number of elements in a list or tuple. The following example introduces both collection types and shows how to access their elements.

```python
word_list = ['I', 'am', 'a', 'list']
word_tuple = ('I', 'am', 'a', 'tuple')

print('List:', word_list)
print('Tuple:', word_tuple)

print(word_list[0], word_tuple[1], 'generated from',
      word_tuple[2], word_tuple[3], 'and', word_list[2], word_list[3])

print('Length of list:', len(word_list))
print('Length of tuple:', len(word_tuple))

print('Type of list:', type(word_list))
print('Type of tuple:', type(word_tuple))
```

What is the difference between a list and a tuple? A list is **mutable**, meaning that its elements can be changed, added, or removed after creation. A tuple is **immutable**, meaning that its elements cannot be replaced, added, or removed. In practice, lists are useful for collections that may change, while tuples are suitable for values that belong together and should remain fixed. The following example demonstrates this difference:

```python
number_list = [1, 2, 3]
number_tuple = (1, 2, 3)

number_list[0] = -1
print('Changed list:', number_list)

# This operation is not allowed and would produce an error:
# number_tuple[0] = -1
```

Because lists are mutable, Python provides several operations for changing their contents. Elements can be replaced, appended, or removed, and lists can be joined to form a new list. The following examples introduce these operations:

```python
values = [1, 2, 3]
print('Initial list:', values)

values[0] = -1
print('After changing the first element:', values)

values.append(10)
print('After appending 10:', values)

# A list may combine different types of values
values = values + ['a', '12', [13, 14]]
print('After concatenating two lists:', values)
print('Last element:', values[-1])

removed_value = values.pop(2)
print('Value removed at index 2:', removed_value)
print('List after pop:', values)

values.remove([13, 14])
print('After removing the value [13, 14]:', values)

del values[2]
print('After deleting the element at index 2:', values)
```

A **slice** selects a range of elements from a list using the notation `[start:stop:step]`. The element at `start` is included, while the element at `stop` is excluded. For example, `[1:4]` selects the elements at indices `1`, `2`, and `3`. Negative indices count backward from the end of the list, with `-1` referring to the last element. If omitted, `start` refers to the beginning, `stop` extends to the end, and `step` has the default value `1`.

```python
values = [11, 12, 13, 14, 15]

print('values =', values)
print('values[0:3] =', values[0:3])      # Indices 0, 1, and 2
print('values[1:3] =', values[1:3])      # Indices 1 and 2
print('values[-1] =', values[-1])        # Last element

print('values[0:4:2] =', values[0:4:2])  # From index 0 to 3, in steps of 2
print('values[0::2] =', values[0::2])    # Every second element
print('values[::-1] =', values[::-1])    # All elements in reverse order
```

The elements of a list or tuple can be assigned to separate variables in a single statement. This operation is called **unpacking**, as illustrated by the following examples:

```python
number_list = [1, 2]
a, b = number_list
print('a =', a, 'and b =', b)

number_tuple = (3, 4)
c, d = number_tuple
print('c =', c, 'and d =', d)
```

A tuple can also be created without parentheses. In Python, it is the commas that define the tuple, as shown in the following example:

```python
number_tuple = 1, 2
print('Tuple:', number_tuple)
```

The `range()` function represents a sequence of integers without storing all its values explicitly. The arguments `start`, `stop`, and `step` control the sequence, with `stop` excluded. A range can be converted into a list or tuple when the individual values are needed.

```python
print('Range:', range(9))
print('Range with start, stop, and step:', range(1, 9, 2))

print('Range converted to a list:', list(range(9)))
print('Range converted to a tuple:', tuple(range(1, 9, 2)))
print('Decreasing range:', list(range(9, 1, -1)))
```

## Boolean Values

A **Boolean value** represents one of two possibilities: `True` or `False`. Boolean values often arise when Python compares two values and are used to decide whether parts of a program should be executed. The following examples introduce some basic comparisons. Note that `=` assigns a value, while `==` compares two values. The two operators look similar but have very different purposes.

```python
a = 1
b = 2

print(a < b)     # Is a smaller than b?
print(a <= b)    # Is a smaller than or equal to b?
print(a == b)    # Are a and b equal?
print(a != b)    # Are a and b different?
```

The function `bool()` converts a value to either `True` or `False`. Numeric zeros and empty objects are converted to `False`; most other values are converted to `True`. The following examples illustrate this rule:

```python
print(bool('a'))              # Non-empty string
print(bool(''))               # Empty string
print(bool(1))                # Non-zero integer
print(bool(0))                # Zero integer
print(bool(0.0))              # Zero floating-point number
print(bool([]))               # Empty list
print(bool([4, 'hello', 1]))  # Non-empty list
```

<a id='sets'></a> 
## Sets

A **set** is an unordered collection in which every element occurs only once. Sets are useful when duplicate values should be removed or when collections should be compared. They also support familiar mathematical operations such as union and intersection. In Python, a set is written using braces `{}`.

```python
number_set = {4, 2, 1, 2, 5, 2}
print('Set:', number_set)  # Repeated values occur only once

set_a = {1, 2, 3}
set_b = {2, 3, 4}
print('Union:', set_a | set_b)             # Elements in either set
print('Intersection:', set_a & set_b)      # Elements in both sets

number_set.add(7)
print('After adding 7:', number_set)
number_set.remove(2)
print('After removing 2:', number_set)
```

The order shown in the output is only one possible order. Unlike lists and tuples, sets do not assign fixed positions to their elements, so an individual element cannot be accessed by an index. The defining properties of a set are uniqueness and membership. The operator `in` can be used to test whether a particular value belongs to a set.

```python
number_set = {1, 2, 4, 5}
print('Contains 2:', 2 in number_set)
print('Contains 3:', 3 in number_set)
```

## Dictionaries

The final collection type introduced in this unit is the **dictionary**. Lists and tuples organize values by position, and sets collect unique elements. A dictionary takes a different approach: it stores information as pairs of **keys** and **values**. Each value is accessed through its key rather than through a numerical index. This makes dictionaries useful for describing objects with clearly named properties, such as a signal with a name, frequency, and duration. A dictionary is written using braces, with a colon separating each key from its value:

```python
sound = {
    'pitch': 'A4',
    'frequency_hz': 440,
    'duration_seconds': 2.0
}

print('Dictionary:', sound)
print('Keys:', list(sound.keys()))
print('Values:', list(sound.values()))
print('Frequency:', sound['frequency_hz'])
```

Each key in a dictionary must be unique. A key can be used to retrieve its associated value, as in `sound['frequency']`.

## Python Type Conversion

Every value in Python has a **type**, which determines how the value is represented and which operations can be applied to it. Two common numerical types are `int` for integers and `float` for floating-point numbers. The function `type()` identifies the type of a value.

```python
integer_number = 1
floating_number = 2.2

print(integer_number, type(integer_number))
print(floating_number, type(floating_number))
```

The functions `float()` and `int()` convert a number from one numerical type to the other. Note that when converting a floating-point number to an integer, Python does not round the value. Instead, it removes the fractional part, moving the value toward zero, as demonstrated by the following examples:

```python
integer_number = 1
floating_number = float(integer_number)
print('Integer converted to float:', floating_number)

positive_float = 2.9
negative_float = -2.9
print('Positive float converted to integer:', int(positive_float))
print('Negative float converted to integer:', int(negative_float))
```

Python also provides the numerical type `complex` for complex numbers. The function `complex(a, b)` constructs a complex number with real part `a` and imaginary part `b`. While the imaginary unit is commonly written as $i$ in mathematics, Python uses `j`.

```python
z = complex(3, 4)

print('Complex number:', z)
print('Type:', type(z))
print('Real part:', z.real)
print('Imaginary part:', z.imag)
print('Magnitude:', abs(z))
```

A complex number cannot be converted directly to `int` or `float` because this would require deciding which information to keep. Instead, one must first select the real part using `.real`, the imaginary part using `.imag`, or the magnitude using `abs()`:

```python
real_integer = int(z.real)
magnitude_float = abs(z)

print('Real part converted to integer:', real_integer)
print('Magnitude as a float:', magnitude_float)
```

## Shallow and Deep Copy Operations

Assignment becomes slightly more subtle when working with mutable objects such as lists, dictionaries, and sets. The statement `b = a` does not copy the object stored under `a`. Instead, it gives the same object a second name, so a change made through either name is visible through both. The operator `is` tests whether two variables refer to the same object. In the following example, `a is b` returns `True`, confirming that no copy was created:

```python
a = [1, 2, 3]
b = a

b[0] = 0
b.append(4)

print('a =', a)
print('b =', b)
print('Do a and b refer to the same object?', a is b)
```

To create copies explicitly, Python provides the [`copy`](https://docs.python.org/3/library/copy.html) module. Two kinds of copies become important for nested objects such as lists containing other lists. A **shallow copy** creates a new outer object but shares the objects nested inside it. A **deep copy** recursively creates independent copies of the outer object and its nested contents. A useful picture is to imagine a list as a box containing smaller boxes. A shallow copy creates a new outer box but reuses the inner boxes. A deep copy creates a new outer box together with new inner boxes. Let us first examine a shallow copy:

```python
import copy

list_a = [[1, 2, 3], [4, 5, 6]]
list_b = copy.copy(list_a)

print('Same outer list?', list_a is list_b)
print('Same first inner list?', list_a[0] is list_b[0])

list_b[0] = [7, 8, 9]
print('After replacing an inner list:')
print('list_a =', list_a)
print('list_b =', list_b)

list_b[1][0] = 0
print('After changing a shared inner list:')
print('list_a =', list_a)
print('list_b =', list_b)
```

The outer lists `list_a` and `list_b` are separate objects. Replacing an entire inner list in `list_b` therefore leaves `list_a` unchanged. However, the second inner list is shared by both outer lists, so changing one of its elements affects both `list_a` and `list_b`. A deep copy removes this connection by copying the nested objects as well:


```python
list_a = [[1, 2, 3], [4, 5, 6]]
list_b = copy.deepcopy(list_a)

print('Same outer list?', list_a is list_b)
print('Same first inner list?', list_a[0] is list_b[0])

list_b[1][0] = 0
print('After changing an element in list_b:')
print('list_a =', list_a)
print('list_b =', list_b)
```

The deep copy creates independent nested lists, so changing `list_b` leaves `list_a` unchanged. As a practical rule, a shallow copy is usually sufficient for a flat collection or a collection containing only immutable values. For nested mutable objects, use a deep copy when the nested contents must also be independent.

<p style="color:#E65100;">
<strong>Note: Copy with Care.</strong> This topic deserves a warning because shared references are a particularly common source of subtle errors in Python. A second name is not a second object: the innocent-looking assignment <code>b = a</code> makes both names refer to the same object, while a shallow copy still shares nested mutable objects. A change in one place may therefore make a surprise appearance somewhere else. This behavior is based on object references, a concept related to pointers in lower-level programming languages, although Python hides the underlying memory management. When independence matters, check object identity with <code>is</code> and choose <code>copy.copy()</code> or <code>copy.deepcopy()</code> deliberately. Further details can be found in the Python documentation on <a href="https://docs.python.org/3/library/copy.html">copy operations</a> and the FAQ entry <a href="https://docs.python.org/3/faq/programming.html#why-did-changing-list-y-also-change-list-x">Why did changing list y also change list x?</a>
