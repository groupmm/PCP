">Background: Why NumPy?</h2>

<p style="color:#1B5E20;">
Python is a flexible general-purpose language, but its basic data structures are not designed specifically for large numerical calculations. Scientific applications often process thousands or millions of related values, such as samples of an audio signal, pixels of an image, or measurements from an experiment. NumPy provides a  <strong>data structure </strong> and a  <strong>collection of operations </strong> designed precisely for this kind of work.
</p>

<p style="color:#1B5E20;">
The central NumPy object is the <strong>array</strong>. An array organizes numerical values in a regular structure with one or more <strong>axes</strong>. A one-dimensional array may represent a sequence or vector, a two-dimensional array may represent a table or matrix, and arrays with three or more axes can represent data such as color images, collections of signals, or simulation results. The number of entries along each axis defines the array's <strong>shape</strong>. For example, an array with shape <code>(2, 3)</code> has two rows and three columns. An array is closely related to what many machine-learning frameworks call a <strong>tensor</strong>. In programming, both terms often refer to <strong>multidimensional collections</strong> of numbers. NumPy uses the term array, while frameworks such as PyTorch use tensor for a similar data structure with additional capabilities, including GPU computation and automatic differentiation. In mathematics, however, the word tensor has a more specific meaning and should not always be identified directly with a multidimensional array.
</p>

<p style="color:#1B5E20;">
A Python list is highly flexible and may contain objects of many different types. A <strong>numerical NumPy array</strong> typically stores all its elements using one fixed data type, such as <code>float64</code> or <code>int32</code>, in a compact data buffer. This homogeneous representation generally requires less memory and allows optimized numerical routines to process the data efficiently. Many arrays are stored contiguously, although slicing, transposition, and similar operations may create noncontiguous <strong>views</strong> of existing data. NumPy also supports <strong>vectorized operations</strong>. Instead of writing a Python loop that processes one element at a time, a calculation is expressed for an entire array. For example, <code>x ** 2</code> squares every element of <code>x</code>, while <code>x + 1</code> adds one to every element. The required loops still exist, but they are carried out internally by optimized compiled implementations. Vectorization makes numerical code shorter, easier to read, and often considerably faster.
</p>

<p style="color:#1B5E20; margin-bottom:0;">
<strong>Why NumPy?</strong> It combines the readability of Python with data structures and compiled operations designed for numerical computation. This makes it possible to move naturally from a mathematical expression to an efficient executable experiment. NumPy has therefore become a foundation of the scientific Python ecosystem and is widely used for signals, images, measurements, simulations, and machine-learning data. Further details can be found in the NumPy introduction <a href="https://numpy.org/doc/stable/user/whatisnumpy.html">What is NumPy?</a>.
</p>

## Importing and Exploring NumPy

NumPy is a large numerical toolbox, and no one needs to memorize every function it contains. A more useful skill is knowing how to import the package, inspect what it provides, and request help when needed. NumPy is already installed as part of the [PCP environment](PCP_01_getstarted.html). By convention, it is imported using the short name `np`:

```python
import numpy as np

print('NumPy version:', np.__version__)
```

The prefix `np` gives access to the objects provided by NumPy. For example, `np.sqrt()` computes square roots, while `np.pi` provides the number $\pi$:

```python
print('Square root of 9:', np.sqrt(9))
print('Pi:', np.pi)
```

The function `dir()` returns a sorted Python list of strings containing the names available for an object. When applied to the NumPy module, `dir(np)` lists the names of its functions, classes, constants, submodules, and other attributes. Since NumPy provides hundreds of such names, printing the complete list is not especially helpful. Instead, the list can be inspected or searched:

```python
numpy_names = dir(np)

print('Number of available names:', len(numpy_names))
print('Is sqrt available?', 'sqrt' in numpy_names)
print('Is pi available?', 'pi' in numpy_names)
print('Is fft available?', 'fft' in numpy_names)
print('Is square_root available?', 'square_root' in numpy_names)
```

Python provides several ways to explore a package. The function `dir(np)` returns the available names, while `help()` displays detailed documentation for a selected object. For example, `help(np.sqrt)` describes the NumPy square-root function. Within a Jupyter notebook, the question-mark syntax `np.sqrt?` provides similar information. Since the resulting documentation may be long, we do not display it here. In practice, `dir()` shows what is available, while `help()` and the searchable [NumPy documentation](https://numpy.org/doc/stable/) explain how individual objects work.

## Creating and Inspecting NumPy Arrays

The central NumPy data structure is the **array**. NumPy arrays are objects of the class `numpy.ndarray`, which can also be written as `np.ndarray` after importing NumPy as `np`. They are commonly created using the function `np.array()`.

An array organizes values along one or more axes and typically stores all elements using the same numerical data type. Four properties are particularly important:

* `shape` gives the number of elements along each axis.
* `ndim` gives the number of axes.
* `size` gives the total number of elements.
* `dtype` gives the numerical type used to store the elements.

The following example creates a **one-dimensional array** from a Python list and displays these properties. As introduced in [Unit 2 on Python type conversion](PCP_02_python.html#type), `type()` identifies the type of the complete Python object. For a NumPy array, it returns `numpy.ndarray`. This should not be confused with `dtype`, which describes the numerical type of the elements stored inside the array.

```python
x = np.array([1, 2, 3, 3])

print('Array:', x)
print('Object type:', type(x))
print('Shape:', x.shape)
print('Number of axes:', x.ndim)
print('Number of elements:', x.size)
print('Data type:', x.dtype)
```

In this example, the shape `(4,)` indicates one axis containing four elements. Note that the **shape** of an array is always represented by a **tuple**. The comma is required for a one-element tuple because Python interprets `(4)` simply as the integer `4`. NumPy also infers the data type from the input values. Here, the result is `int64`, although the default integer type may depend on the computing platform, such as the processor architecture and NumPy build. When a particular type is required, it can be specified explicitly, for example with `np.array([1, 2, 3, 3], dtype=np.int64)`.

In linear algebra, a rectangular arrangement of numbers in rows and columns is called a **matrix**. Such a matrix can be represented in NumPy by a **two-dimensional array**, where the first axis corresponds to the rows and the second axis to the columns. In the following example, the data type is explicitly set to `float64`, so NumPy stores the integer input values as floating-point numbers.

```python
x = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float64)

print('Two-dimensional array:', x, sep='\n')
print('Shape:', x.shape)
print('Number of axes:', x.ndim)
print('Number of elements:', x.size)
print('Data type:', x.dtype)
```

The shape `(2, 3)` indicates two rows and three columns. More generally, each entry of the shape tuple gives the size of one axis. NumPy provides several convenient functions for creating arrays with particular values or structures:

* `np.zeros(shape)` creates an array filled with zeros.
* `np.ones(shape)` creates an array filled with ones.
* `np.arange(start, stop, step)` creates values separated by a fixed step, with `stop` excluded.
* `np.linspace(start, stop, number)` creates a specified number of evenly spaced values and includes both endpoints by default.
* `np.eye(n)` creates an \(n \times n\) identity matrix, with ones on the main diagonal and zeros elsewhere.

Functions such as `np.zeros()`, `np.ones()`, `np.linspace()`, and `np.eye()` create floating-point arrays by default. This is why values such as `0.` and `1.` contain a decimal point when printed. As discussed above, a different data type can be requested explicitly using `dtype`. The function `np.arange()` infers an integer type when all its arguments are integers.


```python
print('Zeros:', np.zeros(3))
print('Integer zeros:', np.zeros(3, dtype=np.int64))
print('Ones:', np.ones((2, 3)), sep='\n')
print('Values with fixed step:', np.arange(2, 8, 2))
print('Five values from 0 to 1:', np.linspace(0, 1, 5))
print('Identity matrix of size 3:', np.eye(3), sep='\n')
```

<a id="reshape"></a>
## Reshaping NumPy Arrays

Reshaping changes how the elements of an array are organized along its axes without adding or removing any values. The total number of elements must therefore remain unchanged. For example, the following one-dimensional array contains 24 values, which can be arranged into several different shapes:

```python
x = np.arange(2 * 3 * 4)
print('x =', x)
print('Shape:', x.shape)

y = x.reshape((3, 8))
print('y =', y, sep='\n')
print('Shape:', y.shape)

z = x.reshape((3, 2, 4))
print('z =', z, sep='\n')
print('Shape:', z.shape)
```

In the previous example, the array `z` has three axes. To access an individual element, one index must be provided for each axis:

```python
print('z[0, 1, 2] =', z[0, 1, 2])
```

NumPy allows one entry of a new shape to be specified as `-1`. This acts as a placeholder whose value is inferred from the number of available elements. For example, an array with six elements can be reshaped to `(3, -1)`, and NumPy determines that the missing value must be `2`.

The shapes `(6,)`, `(6, 1)`, and `(1, 6)` describe different arrays. The first has one axis, while the other two have two axes. In `(6, 1)`, the second axis contains one element; in `(1, 6)`, the first axis contains one element. The position of this singleton axis affects later array and matrix operations.

```python
array_1d = np.arange(6)

array_3x2 = array_1d.reshape((3, -1))
array_6x1 = array_1d.reshape((-1, 1))
array_1x6 = array_1d.reshape((1, -1))

print(f'array_1d:  shape = {array_1d.shape}, axes = {array_1d.ndim}')
print(f'array_3x2: shape = {array_3x2.shape}, axes = {array_3x2.ndim}')
print(f'array_6x1: shape = {array_6x1.shape}, axes = {array_6x1.ndim}')
print(f'array_1x6: shape = {array_1x6.shape}, axes = {array_1x6.ndim}')
```

## Operations on NumPy Arrays

As discussed in the [background section](#background_numpy), NumPy supports **vectorized operations**. In other words, a single expression can describe a calculation for an entire array without requiring an explicit Python loop over its elements. Many familiar arithmetic and comparison operators can be used in this way. In the following examples, NumPy applies each operation element by element and returns a new array containing the results:

```python
x = np.arange(5)

print('x =', x)
print('x + 1 =', x + 1)
print('x * 2 =', x * 2)
print('x * x =', x * x)
print('x ** 3 =', x ** 3)
print('x / 4 =', x / 4)
print('x // 4 =', x // 4)
print('x > 2 =', x > 2)
```

An element-wise operation does not always require two arrays with exactly the same shape. NumPy may automatically adapt a smaller value or array to the required shape, a mechanism called **broadcasting**. The simplest example is `x + 1`: NumPy conceptually applies the scalar value `1` to every element of `x` without creating an equally shaped array of ones. Broadcasting can also combine arrays of different but compatible shapes. In the following example, an array with shape `(3,)` is added to an array with shape `(2, 3)`. NumPy applies the three values to both rows:

```python
x = np.arange(6).reshape((2, 3))
offset = np.array([10, 20, 30])

print('x =', x, sep='\n')
print('offset =', offset)
print('x + offset =', x + offset, sep='\n')
```

An important point is that the same operator may behave differently for NumPy arrays and Python lists. For arrays with compatible shapes, `+` performs element-wise addition. For lists, the same operator concatenates the two collections:

```python
array_a = np.array([0, 1, 2, 3])
array_b = np.array([0, 1, 2, 3])
print('Array addition:', array_a + array_b)

list_a = [0, 1, 2, 3]
list_b = [0, 1, 2, 3]
print('List concatenation:', list_a + list_b)
```

When two-dimensional arrays represent matrices, the word multiplication can refer to two different operations. The operator `*` multiplies corresponding elements and usually preserves the array shape. The operator `@` computes the mathematical matrix product by combining rows of the first matrix with columns of the second. Element-wise multiplication requires compatible shapes, while matrix multiplication requires the inner dimensions to match:

```python
a = np.arange(4).reshape((2, 2))
b = 2 * np.ones((2, 2), dtype=np.int64)

print('Matrix a:', a, sep='\n')
print('Matrix b:', b, sep='\n')

print('Element-wise sum a + b:', a + b, sep='\n')
print('Element-wise product a * b:', a * b, sep='\n')
print('Matrix product a @ b:', a @ b, sep='\n')
```

Many NumPy operations can combine several elements into a smaller result. Such operations are called **reductions**. As an example, consider the method `sum()`, which adds all elements and returns a single value when no axis is specified. For a two-dimensional array, the summation can also be performed along a selected axis. Summing along `axis=0` combines the rows and produces one value for each column, while summing along `axis=1` combines the columns and produces one value for each row. In each case, the selected axis disappears from the result, as illustrated by the following example:

```python
x = np.arange(6).reshape((2, 3))

sum_axis_0 = x.sum(axis=0)
sum_axis_1 = x.sum(axis=1)

print('Array x:', x, sep='\n')
print('Shape of x:', x.shape)
print('Sum of all elements:', x.sum())
print('Sum along axis 0:', sum_axis_0, 'Shape:', sum_axis_0.shape)
print('Sum along axis 1:', sum_axis_1, 'Shape:', sum_axis_1.shape)
```

NumPy follows the zero-based indexing and slicing conventions introduced for [Python lists](PCP_02_python.html#lists). Index `0` refers to the first element, and the `stop` value of a slice is excluded. For a two-dimensional array, the first index refers to the rows and the second to the columns. A colon `:` selects all elements along the corresponding axis.

A particularly useful form of indexing is based on comparisons. An expression such as `x > 1` creates a Boolean array with the same shape as `x`, containing `True` wherever the condition is satisfied. This Boolean array is called a **mask** and can be used to select the corresponding elements. The following example illustrates indexing, slicing, and masking:

```python
x = np.arange(6).reshape((2, 3))

print('Array x:', x, sep='\n')
print('Element x[1, 1]:', x[1, 1])
print('Slice of first row x[0, 1:3]:', x[0, 1:3])
print('Second row x[1, :]:', x[1, :])
print('Second column x[:, 1]:', x[:, 1])

mask = x > 1
print('Boolean mask x > 1:', mask, sep='\n')
print('Elements selected by the mask:', x[mask])
```

## NumPy Data Types and Conversion

As discussed in [Unit 2 on Python type conversion](PCP_02_python.html#type), Python provides the numerical types `int`, `float`, and `complex`. NumPy provides more specific numerical types whose names often indicate the amount of storage they use. For example, `np.int32` represents a signed integer using 32 bits, while `np.float64` represents a floating-point number using 64 bits.

For an array `x`, the expression `type(x)` identifies the complete object as a NumPy array, while `x.dtype` describes the numerical type of its elements. A particular data type can be selected when an array is created using the argument `dtype`. An existing array can be converted using the method `astype()`, which returns a new array and leaves the original unchanged. The following examples illustrate these conversions:

```python
integer_array = np.array([-1, 2, 3], dtype=np.int64)
floating_array = integer_array.astype(np.float64)
complex_array = integer_array.astype(np.complex128)

print(f'Integer array:        {integer_array}; dtype: {integer_array.dtype}')
print(f'Floating-point array: {floating_array}; dtype: {floating_array.dtype}')
print(f'Complex array:        {complex_array}; dtype: {complex_array.dtype}')
```

The data type determines not only how numbers are stored, but also which results can be represented. For example, the square root of a negative number is not defined within the real numbers. When `np.sqrt()` is applied to a real-valued array containing negative values, NumPy returns `nan` for these entries and issues a `RuntimeWarning`. Here, `nan` means that no valid real-valued result is available.

```python
real_values = np.arange(-2, 2, dtype=np.float64)
real_roots = np.sqrt(real_values)

print(f'Real values: {real_values}; dtype: {real_values.dtype}')
print(f'Square roots: {real_roots}; dtype: {real_roots.dtype}')
```

The warning in this example is expected: NumPy is drawing attention to operations that have no valid result in the real number system. This illustrates why choosing a suitable data type matters, since NumPy can only return results within the numerical domain represented by the array. Converting the input to a complex data type extends this domain, allowing the negative values to have valid complex square roots:

```python
complex_values = real_values.astype(np.complex128)
complex_roots = np.sqrt(complex_values)

print(f'Complex values: {complex_values}; dtype: {complex_values.dtype}')
print(f'Square roots: {complex_roots}; dtype: {complex_roots.dtype}')
```

## Mathematical Constants and Special Values

Numerical calculations often involve well-known mathematical constants or results that cannot be represented as ordinary finite numbers. NumPy provides convenient names for both situations. The first two values below are mathematical constants, while `inf` and `nan` are special floating-point values used to mark exceptional numerical results:

* `np.pi` represents Archimedes' constant $\pi$, the ratio of a circle's circumference to its diameter. It appears throughout geometry, trigonometry, and signal processing.
* `np.e` represents Euler's number $e$, the base of the natural logarithm. It plays a central role in exponential growth, decay, and oscillation.
* `np.inf` and `-np.inf` represent positive and negative infinity. They may arise, for example, when a nonzero floating-point number is divided by zero.
* `np.nan` represents **Not a Number**. It indicates that a numerical result is undefined or unavailable, as in the real-valued square root of a negative number.

```python
print('Pi:', np.pi)
print('Euler’s number:', np.e)
print('Positive infinity:', np.inf)
print('Negative infinity:', -np.inf)
print('Not a Number:', np.nan)
```

Unlike the mathematical constants `np.pi` and `np.e`, the special values `np.inf` and `np.nan` may arise automatically during a calculation. In NumPy floating-point array operations, dividing a positive number by zero produces `inf`, dividing a negative number by zero produces `-inf`, and dividing zero by zero produces `nan`. NumPy continues the calculation while marking the exceptional positions, as illustrated by the following example:

```python
print('10 + inf =', 10 + np.inf)
print('10 + (-inf) =', 10 + (-np.inf))
print('10 + nan =', 10 + np.nan)
print('inf + nan =', np.inf + np.nan)
print('inf - inf =', np.inf - np.inf)
```

NumPy provides Boolean tests for identifying special values in an array. The functions `np.isinf()`, `np.isnan()`, and `np.isfinite()` identify infinite, `nan`, and finite values, respectively. Each function returns a Boolean mask with the same shape as the input array. As with the masks introduced earlier, this mask can be used directly to select the corresponding values.

In the following example, several divisions are performed deliberately to generate negative infinity, `nan`, positive infinity, and a finite result. Since the exceptional results are expected, [`np.errstate()`](https://numpy.org/doc/stable/reference/generated/numpy.errstate.html) temporarily suppresses the resulting warnings:

```python
numerator = np.array([-2.0, 0.0, 2.0, 4.0, 6.0, 8.0])
denominator = np.array([0.0, 0.0, 0.0, 0.0, 2.0, 2.0])

with np.errstate(divide='ignore', invalid='ignore'):
    result = numerator / denominator

mask_inf = np.isinf(result)
mask_nan = np.isnan(result)
mask_finite = np.isfinite(result)

print('Result:', result)
print('Infinity mask:', mask_inf)
print('NaN mask:', mask_nan)
print('Finite-value mask:', mask_finite)

print('Infinite values:', result[mask_inf])
print('NaN values:', result[mask_nan])
print('Finite values:', result[mask_finite])
```

<h2 style="color:#1B5E20;">Background: Matrices and Basic Operations</h2>

<p style="color:#1B5E20;">
Let $N,M\in\mathbb{N}$ be two positive integers. An $(N\times M)$-matrix $\mathbf{A}$ is a rectangular array of entries arranged in <strong>rows</strong> and <strong>columns</strong>. For example, if entries are real numbers, one also writes $\mathbf{A}\in\mathbb{R}^{N\times M}$. Let $a_{nm}\in\mathbb{R}$ denote the entry of $\mathbf{A}$ being in the $n^{\mathrm{th}}$ row and $m^{\mathrm{th}}$ column for $n\in[1:N]=\{1,2,...,N\}$ and $m\in[1:M]$. Then, one also often writes $\mathbf{A}=[a_{nm}]$. A <strong>vector</strong> is a matrix where either $N=1$ (then called <strong>row vector</strong>) or $M=1$ (then called <strong>column vector</strong>).
<br><br>
When computing with vectors and matrices, one has to pay attention that the dimensions $N$ and $M$ of the operands match properly. For example, for a <strong>matrix summation</strong> or <strong>matrix subtraction</strong>, the operands must be of same dimensions or one of them needs to be a scalar (in this case the operations are applied per entry). The multiplication of matrices refers to a <strong>matrix multiplication</strong>. For an $(N\times M)$-matrix $\mathbf{A} = [a_{nm}]$ and a $(M\times P)$-matrix $\mathbf{B} = [b_{mp}]$, the product matrix $\mathbf{C} = \mathbf{A}\mathbf{B}$ with entries $\mathbf{C}=[c_{np}]$ is defined by $c_{np} = \sum_{m=1}^M a_{nm}b_{mp}$ for $n\in[1:N]$ and $p\in[1:P]$. In other words, the entry $c_{np}$ is the <strong>inner product</strong> (sometimes also called <strong>scalar product</strong>) of $n^{\mathrm{th}}$ row of $\mathbf{A}$ and $p^{\mathrm{th}}$ column of $\mathbf{B}$. This calculation rule is illustrated by the following figure:
