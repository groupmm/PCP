<h2 style="color:#1B5E20; margin-top:0;">
Background: Algorithmic Thinking
</h2>

<p style="color:#1B5E20;">
Solving a problem with a computer usually begins long before the first line of code is written. One must identify the available inputs, the desired result, the necessary steps, the choices that may arise, and the point at which the process should stop. Organizing a problem in this systematic way is known as <strong>algorithmic thinking</strong>. The resulting <strong>algorithm</strong> is a precise and finite method for solving the problem. In this respect, an algorithm resembles a <strong>cooking recipe</strong>: it lists the ingredients, orders the necessary steps, and indicates when an action should be repeated or the process is complete. A computer is less willing to improvise than a cook, so every computational step must be stated without ambiguity.
</p>

<p style="color:#1B5E20;">
Algorithmic thinking is one of the foundations of programming. An algorithm describes a solution independently of a particular programming language, while a <strong>program</strong> expresses that solution in instructions a computer can execute. Programming elements give concrete form to the structure of an algorithm:
</p>

<ul style="color:#1B5E20;">
<li><strong>Statements</strong> express individual operations.</li>
<li><strong>Conditions</strong> choose between alternative paths.</li>
<li><strong>Loops</strong> repeat operations.</li>
<li><strong>Functions</strong> collect related steps into reusable units with defined inputs and outputs.</li>
</ul>

<p style="color:#1B5E20;">
Programming is therefore more than writing commands in the correct syntax. It involves turning an idea into a form in which every step, decision, and stopping condition is clear. Code can even serve as a test of understanding: if a method cannot yet be expressed precisely, some part of the underlying idea may still need attention.
</p>

<p style="color:#1B5E20;">
Consider the problem of finding a root of a continuous mathematical function. Starting with an interval containing a root, one computes its midpoint and determines in which half the function changes sign. The other half is discarded, and the process is repeated until the remaining interval is sufficiently small. This <strong>bisection method</strong> combines calculations, decisions, repetition, and a stopping condition. A Python function can package the complete algorithm so that it works with different mathematical functions and intervals. This example illustrates the path from an idea to an executable solution. Mathematics provides the underlying method, algorithmic thinking organizes it into precise steps, control flow determines how those steps are executed, and functions make the resulting program reusable. Later in this unit, we will complete this path by implementing the bisection method ourselves.
</p>

## Conditions and Branching

Python normally executes statements one after another, from top to bottom. Sometimes, however, the next step should depend on the available data. A program may need to distinguish positive from negative numbers, check whether a value lies within an interval, or respond differently to valid and invalid input. Choosing between such alternative paths is called **branching**. The choice is guided by a **condition**, an expression that evaluates to one of the Boolean values `True` or `False`, as introduced in [Unit 2](PCP_02_python.html#boolean). Conditions are often constructed using comparisons such as `x < 0`, `x == 0`, or `x >= 10`. Python uses three keywords to express conditional branches:

* `if` starts the first branch and tests a condition.
* `elif` combines the ideas of `else` and `if`: it tests another condition when all preceding conditions were `False`.
* `else` provides a final branch when none of the preceding conditions were satisfied.

Each branch header ends with a colon `:`. The statements belonging to that branch appear on the following lines and are **indented**, usually by four spaces. Indentation is part of Python's syntax and identifies a block of statements that belong together. The following example uses three branches to classify a number:

```python
n = -2

if n < 0:
    print('Negative')
elif n > 0:
    print('Positive')
else:
    print('Zero')
```

Python evaluates the conditions from top to bottom. In this example, `n < 0` is `True`, so Python executes the first branch and skips the remaining ones. Try changing `n` to a positive number and then to zero to explore the other paths. The `elif` and `else` branches are optional, so an `if` statement may also consist of a single condition and its associated block.

<p style="color:#E65100;">
<strong>Note: Indentation Is Syntax.</strong> In Python, indentation does more than improve readability. It determines which statements belong to a condition, loop, or function. Indentation is therefore required, although the language does not prescribe exactly four spaces. The widely followed <a href="https://peps.python.org/pep-0008/#indentation">PEP 8 style guide</a> recommends <strong>four spaces</strong> for each indentation level. Statements in the same block must be aligned consistently, and nested blocks must be indented further. Missing, unexpected, or inconsistent indentation may produce an <code>IndentationError</code>. Tabs are technically permitted, but spaces are preferred, and mixing tabs with spaces may produce a <code>TabError</code>. Even when no error occurs, a misplaced space can change the meaning of a program.
</p>
</div>

A single comparison is not always enough to make a decision. To check whether a number lies inside an interval, for example, we must test both its lower and upper bounds. Python combines or modifies such Boolean expressions using three **logical operators**:

* `and` returns `True` only when both conditions are `True`.
* `or` returns `True` when at least one of the conditions is `True`.
* `not` reverses a condition, changing `True` to `False` and `False` to `True`.

The following example uses `and` because the number must satisfy both interval boundaries at the same time:

```python
x = 7
lower = 0
upper = 10

if x >= lower and x <= upper:
    print('x lies inside the interval')
else:
    print('x lies outside the interval')
```

Python also allows related comparisons to be **chained**. Instead of writing two separate comparisons joined by `and`, an interval condition can be written as a single expression that closely resembles mathematical notation:

```python
if lower <= x <= upper:
    print('x lies inside the interval')
else:
    print('x lies outside the interval')
```

<p style="color:#E65100;">
<strong>Note: One Condition or an Array of Conditions?</strong> An individual comparison produces one Boolean value, while a NumPy comparison may produce an entire array of Boolean values. Python therefore uses different operators in these two situations:
</p>

<ul style="color:#E65100;">
<li>
For <strong>individual Boolean expressions</strong>, use <code>and</code>, <code>or</code>, and <code>not</code>. For example, <code>x &gt; 0 and x &lt; 10</code> checks two conditions for a single value. The chained comparison <code>0 &lt; x &lt; 10</code> is equivalent.
</li>

<li>
For <strong>NumPy Boolean arrays</strong>, use <code>&amp;</code>, <code>|</code>, and <code>~</code> to process the values element by element. The operator <code>&amp;</code> means and, <code>|</code> means or, and <code>~</code> reverses each Boolean value. Using <code>and</code>, <code>or</code>, or <code>not</code> with a multi-element array normally produces an error because Python expects a single truth value.
</li>
</ul>

<p style="color:#E65100; margin-bottom:0;">
Always place parentheses around NumPy comparisons. Write <code>(x &gt; 0) &amp; (x &lt; 10)</code>, not <code>x &gt; 0 &amp; x &lt; 10</code>. Without the parentheses, Python groups the operations differently and does not perform the intended comparison.
</p>

## Loops and Iteration

Many computational tasks require the same operation to be performed several times. A **loop** repeats a block of statements, and each repetition is called an **iteration**. Python provides two main kinds of loops:

* A `for` loop processes the elements of a collection or sequence one after another.
* A `while` loop repeats a block of statements as long as a condition remains `True`.

A `for` loop visits the elements of an **iterable**, such as a range, list, tuple, or string. During each iteration, the current element is assigned to a loop variable. The loop header ends with a colon `:`, and the statements belonging to the loop are indented. The following example processes the numbers generated by `range(4)`. Recall that the stopping value is excluded:

```python
for number in range(4):
    print(f'{number} squared is {number**2}')
```

A `for` loop can process many kinds of iterable objects. The loop variable receives one element at a time, regardless of whether those elements are numbers, strings, or other Python objects:

```python
for item in ['a', 2, 'hello']:
    print('List item:', item)

for character in 'PCP':
    print('Character:', character)
```

A `while` loop repeats a block as long as a condition remains `True`. It is useful when the number of iterations is not known in advance, such as when searching for a result or refining an approximation. The condition is checked before each iteration, so the loop may run many times, not at all, or forever if nothing inside the loop gives it a reason to stop:

```python
number = 0

while number < 4:
    print('Current number:', number)
    number += 1
```

  background-color: #FFE0B2A0;

<p style="color:#E65100;">
<strong>Note: Every <code>while</code> Loop Must Make Progress.</strong> A <code>while</code> loop continues until its condition becomes <code>False</code> or a <code>break</code> statement is reached. If nothing inside the loop changes the condition, the loop may continue indefinitely. When writing a <code>while</code> loop, identify which variable controls the condition and check that each iteration updates it in the intended direction. If a loop runs indefinitely in a Jupyter notebook, its execution can be interrupted using the stop button or the menu command <code>Kernel &gt; Interrupt Kernel</code>.
</p>
</div>

Two additional statements provide more control over loop execution. The statement `break` ends the loop immediately. This is useful when the desired result has already been found and no further iterations are needed. In the following example, the loop stops at the first number whose square is greater than `20`:

```python
for number in range(10):
    if number**2 > 20:
        print('First number found:', number)
        break
```

The statement `continue` skips the remaining statements in the current iteration and proceeds directly with the next one. The following example skips all even numbers and prints only the odd ones:

```python
for number in range(6):
    if number % 2 == 0:
        continue
    print('Odd number:', number)
```

The difference is important: `break` ends the complete loop, while `continue` skips only the remainder of the current iteration. When loops are nested, either statement affects only the **innermost loop** containing it.

## Defining Python Functions

In programming, a **function** is a named and reusable block of statements that performs a particular task. It may receive input, process that input, and produce a result. Functions help organize larger programs into manageable parts, avoid repeated code, and give meaningful names to individual computations.

In Python, a function definition begins with the keyword `def`, followed by the function name and parentheses `()`. Any input parameters are listed inside the parentheses, and a colon `:` marks the beginning of the indented function body. The keyword `return` can be used to send a result back to the part of the program that requested the computation. The following example defines a simple function:

```python
def square(number):
    result = number**2
    return result
```

Defining a function makes it available to Python, but the statements in its body are not executed immediately. They run only when the function is **called** by writing its name followed by parentheses. In the definition above, `number` is a **parameter** representing the function’s input. In the call `square(5)`, the value `5` is the **argument** supplied for this parameter. The function computes the square and returns `25` to the calling statement. Once defined, the function can be called repeatedly with different arguments:

```python
print('Square of 2:', square(2))
print('Square of 3:', square(3))
print('Square of 4:', square(4))
```

Parameters and names assigned inside a function normally have **local scope**. They can be used inside the function but cannot be accessed by those names from outside it. For example, `result` is local to `square()`, while the value returned by the function can be stored under a different name by the caller. Local scope keeps intermediate calculations contained and reduces interference with names used elsewhere in the program.

**Function names** follow the same rules as variable names. The common Python convention is to use lowercase letters and underscores, as in `compute_mean` or `find_root`. A descriptive name helps readers recognize the purpose of a function before studying its implementation.

A function body may contain conditions, loops, and calls to other functions. A function should also document its purpose, expected input, and output. This information is placed in a **docstring**, a string that appears as the first statement of the function body. According to the [Python Docstring Conventions](https://peps.python.org/pep-0257/), docstrings are normally enclosed in triple double quotes.

The following example brings these ideas together. It contains a loop and a condition, calls the previously defined function `square()`, and documents its behavior with a docstring. The parameters and intermediate names `numbers`, `total`, and `number` are local to `sum_squares_of_positives()`:

```python
def sum_squares_of_positives(numbers):
    """Return the sum of the squares of all positive input values.

    Args:
        numbers: Collection of numerical values.

    Returns:
        Sum of the squares of the positive values.
    """
    total = 0

    for number in numbers:
        if number > 0:
            total += square(number)

    return total

values = [-2, 1, 3]
result = sum_squares_of_positives(values)

print('Sum of squares of positive numbers:', result)
```

Notice that the variable `result` defined outside the functions is separate from the local variable with the same name inside `square()`. The two names belong to different scopes and therefore do not interfere with each other.

The built-in function `help()` displays a function’s signature and docstring:

```python
help(sum_squares_of_positives)
```

## Arguments and Return Values

Functions can have several parameters. Some may be **required**, while others may have **default values** that are used when no corresponding argument is supplied. The following function has one required parameter, `a`, and two parameters with default values, `b` and `c`:

```python
def add(a, b=0, c=0):
    """Return the sum of up to three numbers."""
    return a + b + c
```

Arguments can be supplied by position or by name:

```python
print('add(5):', add(5))
print('add(5, 2, 1):', add(5, 2, 1))
print('add(5, c=4):', add(5, c=4))
```

In the first call, only the required argument is supplied, so `b` and `c` keep their default values. In the second call, all three arguments are matched to parameters by their positions. In the third call, the positional argument `5` is assigned to `a`, while the keyword argument `c=4` replaces the default value of `c`. The main rules used in this course are:

* Every required parameter must receive an argument.
* Positional arguments are matched to parameters in their given order.
* Keyword arguments are matched using their parameter names.
* A supplied argument replaces the corresponding default value.
* In the simple function definitions used here, required parameters must precede parameters with default values.
* In a function call, positional arguments must normally precede keyword arguments.

A `return` statement ends the current function call and sends a value back to the caller. Sometimes, however, a function does not produce an ordinary result. Python uses the special value `None` to represent the absence of a value. It is different from the number `0`, the Boolean value `False`, and the string `'None'`. A function may return `None` explicitly, but it also returns `None` automatically if it reaches the end without executing a `return` statement. The expression `result is None` can be used to check for this special value. In the following example, `None` indicates that the function did not compute a square because the input was not positive:

```python
def square_if_positive(number):
    """Return the square of a positive number."""
    if number > 0:
        return number**2
    return None

result_positive = square_if_positive(3)
result_negative = square_if_positive(-3)

print('Positive input:', result_positive)
print('Negative input:', result_negative)
print('Negative result is None:', result_negative is None)
```

<p style="color:#E65100; margin-bottom:0;">
<strong>Note: None Is a Value, but Not a Number.</strong> The special value <code>None</code> represents the absence of an ordinary result. A function may return it deliberately, but Python also returns <code>None</code> automatically when the function reaches its end without executing a <code>return</code> statement. Although <code>None</code> can be stored in a variable, returned by a function, or placed in a collection, it cannot normally take part in arithmetic. An expression such as <code>None + 1</code> therefore produces a <code>TypeError</code>. Do not confuse <code>None</code> with <code>0</code>, <code>False</code>, or an empty string, since all of these have different meanings. To check for this special value, use <code>x is None</code>. If <code>None</code> makes an unexpected appearance, a function may have quietly reached its end without returning the intended result. In that case, the missing value is often pointing directly toward a missing <code>return</code>.

A function may also return several values at once. Python combines these values into a tuple, which can either be stored as a single object or unpacked directly into separate variables:

```python
def add_and_difference(a, b):
    """Return the sum and difference of two numbers."""
    return a + b, a - b

result = add_and_difference(3, 5)
print('Returned tuple:', result)
print('Type:', type(result))

sum_value, difference = add_and_difference(3, 5)
print('Sum:', sum_value)
print('Difference:', difference)
```

## Algorithmic Efficiency and Runtime

The [background discussion on algorithmic thinking](#background_algorithm) emphasized that a problem can often be translated into a sequence of precise computational steps. However, finding a correct sequence is only part of the task. The same problem may have several correct solutions that require very different amounts of work. **Algorithmic efficiency** asks how the required time or memory grows as the input becomes larger. This distinction is important because computational performance does not depend only on processor speed, GPUs, or other hardware. A faster computer can execute a given procedure more quickly, but a better algorithm may avoid much of the work altogether. Before asking how fast a computer can perform the steps, it is therefore worth asking whether all those steps are necessary. As a simple example, consider the sum

$$
S(N)=\sum_{n=1}^{N}n=1+2+\cdots+N,
$$

where $n$ is a positive integer. A direct implementation follows this expression literally and adds the numbers one at a time:

```python
def sum_integers_loop(N):
    """Return the sum of the integers from 1 to N using a loop."""
    total = 0

    for n in range(1, N + 1):
        total += n

    return total
```

The following `while` loop evaluates the function for increasing values of `n` and stops before the sum exceeds `30`. The result is computed once during each iteration and stored in `current_sum`:

```python
N = 1
current_sum = sum_integers_loop(N)

while current_sum <= 30:
    print(f'N = {N:2d}, sum = {current_sum:2d}')
    N += 1
    current_sum = sum_integers_loop(N)
```

The direct loop is easy to understand, but it is not the only possible solution. NumPy can generate and sum the complete sequence using optimized numerical operations. Mathematics offers an even shorter route through the closed formula

$$
S(N)=\frac{N(N+1)}{2}.
$$

The following code implements these two alternatives and verifies that all three functions return the same result:

```python
def sum_integers_numpy(N):
    """Return the sum of the integers from 1 to N using NumPy."""
    return np.sum(np.arange(1, N + 1))

def sum_integers_formula(N):
    """Return the sum of the integers from 1 to N using a formula."""
    return N * (N + 1) // 2

N = 1000

result_loop = sum_integers_loop(N)
result_numpy = sum_integers_loop(N)
result_formula = sum_integers_loop(N)

print('Python loop:      ', result_loop)
print('NumPy operation:  ', result_numpy)
print('Closed formula:   ', result_formula)
print('All results agree:', result_loop == result_numpy == result_formula)
```

  padding-top:0;
">Background: Gauss, Complexity, and a Better Idea</h2>

<p style="color:#1B5E20;">
A famous story, which may be partly legendary, tells how the young mathematician Carl Friedrich Gauss was asked to add the integers from $1$ to $100$. Instead of adding them one after another, he recognized a useful pattern. Pairing the first and last numbers gives $1+100=101$, pairing the second and second-to-last gives $2+99=101$, and the same result continues for every pair. There are $50$ such pairs, so the sum is $50\cdot101=5050$.
</p>

<p style="color:#1B5E20;">
The same reasoning works for a general upper limit $N$. Writing the sequence once in ascending order and once in descending order gives
</p>

<p style="color:#1B5E20; text-align:center;">
$S(N)=1+2+\cdots+(N-1)+N,$
<br>
$S(N)=N+(N-1)+\cdots+2+1.$
</p>

<p style="color:#1B5E20;">
Adding corresponding entries produces $N$ copies of $N+1$. Therefore,
</p>

<p style="color:#1B5E20; text-align:center;">
$2S(N)=N(N+1)
\qquad\Longrightarrow\qquad
S(N)=\dfrac{N(N+1)}{2}.$
</p>

<p style="color:#1B5E20;">
This is more than an elegant mathematical shortcut. It illustrates a fundamental idea in computer science: two correct methods may require very different amounts of work. A loop performs approximately $N$ additions, so its computational effort grows in proportion to $N$. This behavior is called <strong>linear complexity</strong> and is written as $\mathcal{O}(N)$. The formula uses only a fixed number of arithmetic operations, regardless of the size of $N$. It therefore has <strong>constant complexity</strong>, written as $\mathcal{O}(1)$.
</p>

<p style="color:#1B5E20;">
The notation $\mathcal{O}(\cdot)$ describes how the required work grows as the input becomes larger. It does not predict an exact runtime in seconds. Those measurements also depend on the computer, software, and implementation. Complexity instead reveals how well a method <strong>scales</strong>. Faster hardware may accelerate both methods, but it does not change the fact that one method performs work that grows with $N$, while the other does not.
</p>

<p style="color:#1B5E20; margin-bottom:0;">
This principle will return in Fourier analysis. A direct computation of a discrete Fourier transform requires approximately $\mathcal{O}(N^2)$ operations, while the fast Fourier transform reduces the work to approximately $\mathcal{O}(N\log N)$. The FFT is fast not merely because it can run on powerful hardware, but because it discovers and reuses structure instead of repeating unnecessary calculations. Gauss and the FFT are very different examples, but they share the same lesson: sometimes the greatest speedup begins with a better idea.
</p>

</div>

The three methods produce the same mathematical result, but they reach it in different ways:

* `sum_integers_loop()` visits and adds all $N$ numbers using a Python loop.
* `sum_integers_loop` also processes all $N$ numbers, but performs the main work using optimized compiled code.
* `sum_integers_loop()` does not generate or visit the sequence. It evaluates only a fixed number of arithmetic operations.

For the first two methods, the amount of work grows approximately in proportion to $N$. This behavior is called **linear complexity** and is written as $\mathcal{O}(N)$. The formula, in contrast, requires a fixed number of operations regardless of $N$. It therefore has **constant complexity**, written as $\mathcal{O}(1)$. NumPy makes the repeated operations faster, while the mathematical formula reduces the amount of work itself.

The [`timeit`](https://docs.python.org/3/library/timeit.html) module can be used to compare the runtime of the three implementations. Each function is executed repeatedly because a single execution may be too brief for a reliable measurement. In expressions such as `lambda: sum_integers_loop(N)`, `lambda` creates a short unnamed function that allows `timeit` to call the calculation when needed. We use this compact syntax only for the runtime measurement; further details are provided in the [Python tutorial on lambda expressions](https://docs.python.org/3/tutorial/controlflow.html#lambda-expressions).

```python
import timeit

N = 100000
num_executions = 100

runtime_loop = timeit.timeit(
    lambda: sum_integers_loop(N),
    number=num_executions
) / num_executions

runtime_numpy = timeit.timeit(
    lambda: sum_integers_numpy(N),
    number=num_executions
) / num_executions

runtime_formula = timeit.timeit(
    lambda: sum_integers_formula(N),
    number=num_executions
) / num_executions

print(f'Python loop:    {runtime_loop * 1000:.6f} ms')
print(f'NumPy:          {runtime_numpy * 1000:.6f} ms')
print(f'Closed formula: {runtime_formula * 1000:.6f} ms')
```

The exact runtimes depend on the computer, software versions, and current system activity, so the overall pattern matters more than the individual measurements. Fast hardware and optimized libraries can accelerate a given calculation, but an even greater improvement may come from choosing a method that requires less work in the first place.

A more thoughtful algorithm may avoid thousands or millions of operations before the hardware performs even the first instruction. The central question is therefore not only <em>How quickly can the computer perform this calculation?</em>, but also <em>How much calculation is actually necessary?</em> Good computational thinking uses hardware wisely, but does not ask hardware to compensate for an inefficient idea.

Let $f:\mathbb{R}\to\mathbb{R}$ be a continuous function and let $a,b\in\mathbb{R}$ be two numbers such that $a<b$ and $f(a)$ and $f(b)$ have opposite signs. Then $f$ has a <strong>root</strong> in the interval $[a,b]$. In other words, there is a number $r\in[a,b]$ such that $f(r)=0$. To find such a root, one can proceed as follows: 
</p>
    
<ul style="color:#006064;">
    <li> Let $c = (a+b)/2$ be the center. If $f(a)=0$, $f(b)=0$, or $f(c)=0$, then we have found a root and we can stop. </li> 
    <li> Otherwise, either $f(a)$ and $f(c)$, or $f(c)$ and $f(b)$ have opposite signs. In the first case, continue with the interval $[a,c]$, otherwise with the interval $[c,b]$.</li> 
    <li> Iterate this interval-halving procedure until the interval size is below a given threshold parameter (e.g., $10^{-5}$) and a sufficient approximation for the root is obtained.</li>
</ul>

Here, <code>f</code> is a Python function realizing a continuous function $f:\mathbb{R}\to\mathbb{R}$. The parameter <code>thresh</code> is the threshold parameter, which should be set to the value $10^{-5}$ by default. The function <code>search_root</code> should check if <code>a < b</code> and if the opposite sign condition for <code>f(a)</code> and <code>f(b)</code> is fulfilled (if not it should return the value <code>np.nan</code>).
